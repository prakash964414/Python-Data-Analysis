```python
#### t test  1 sample test
from scipy.stats import ttest_1samp

marks = df["Marks"]
t_stat, p_value = ttest_1samp(marks,popmean = 80)
print("T-Statistics of the marks:", t_stat)
print("P-Value:", p_value)


### (95% confienace interval = 0.05)
if p_value>0.05:
    print ("Accept the null Hypothesis")
else:
    print("reject the Null  Hypothesis")

    
```
